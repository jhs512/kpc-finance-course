## 1. 아나콘다 설치 및 Jupyter Notebook 실행 환경 구성

### 아나콘다를 짧게 이해하고 uv로 실습하기

공식 커리큘럼의 항목은 **「아나콘다 설치 및 Jupyter Notebook 실행 환경 구성」**입니다. 아나콘다는 Python과 분석 도구를 묶어 제공하는 배포판이고 conda는 환경·패키지를 관리합니다. 아나콘다 설치 방식은 [공식 설치 안내](https://www.anaconda.com/docs/getting-started/anaconda/install/windows-gui-install)로 짧게 소개합니다. 이번 수업의 설치 실습은 사용자가 선택한 현대화 방안에 따라 **uv → Python → 프로젝트 환경 → Jupyter Notebook**으로 진행합니다. 공식 문구와 실제 설치 도구의 차이를 구분합니다.

이 수업에서 필요한 것은 Python, 분석 패키지, Notebook 실행 환경입니다. uv는 Python을 준비하고 폴더별 `.venv`와 의존성을 관리할 수 있어 이 구성에 아나콘다를 추가로 설치할 필요가 없습니다. uv 자체가 Python이나 Notebook을 대신 실행하는 언어는 아닙니다. [uv의 Python 관리](https://docs.astral.sh/uv/guides/install-python/)를 참고합니다. conda의 다양한 비 Python 도구까지 모두 같다고 설명하지 않습니다.

| 도구 | 이번 수업에서 맡는 일 |
| --- | --- |
| uv | Python 준비, 프로젝트 환경·패키지 관리, 환경에 맞는 명령 실행 |
| Python | 계산·조건문·반복문 실행 |
| `.venv` | 이 수업의 패키지를 다른 프로젝트와 분리 |
| Jupyter Notebook | 설명·코드·출력을 셀로 묶어 작성 |
| Kernel | 선택한 Python으로 코드 셀 실행, 변수 상태 유지 |

### Windows에서 uv 준비

**PowerShell**을 열고 다음 명령을 실행합니다. 이미 uv가 있으면 설치는 생략하고 버전 확인부터 합니다.

```powershell
winget install --id=astral-sh.uv -e
```

설치 후 PowerShell을 새로 열고 `uv --version`을 실행합니다. WinGet을 쓸 수 없는 PC와 macOS·Linux는 [uv 공식 설치 문서](https://docs.astral.sh/uv/getting-started/installation/)의 운영체제별 방법을 따릅니다. 이 설치 단계는 수업 전에 마치고, 수업에서는 환경 생성과 첫 셀에 시간을 씁니다.

### 배포한 수업 폴더로 시작하기 — 수강생 기본 경로

강사가 배포한 수업 폴더를 열고, 그 안에 `pyproject.toml`, `uv.lock`, `.python-version`, `raw`가 있는지 확인합니다. PowerShell에서 **그 폴더로 이동한 뒤** 실행합니다. 이미 프로젝트 파일이 있으므로 `uv init`은 다시 실행하지 않습니다.

```powershell
uv sync --locked
uv run --locked jupyter notebook
```

`uv sync --locked`는 배포한 잠금 파일에 맞춰 환경을 준비합니다. 프로젝트 설정과 잠금 파일이 맞지 않으면 임의로 잠금 파일을 바꾸는 대신 오류를 내므로 강사가 배포 자료를 확인합니다. `uv run`을 쓰면 PowerShell에서 `.venv`를 수동 활성화할 필요가 없습니다. Python·패키지를 처음 받는 경우 네트워크와 설치 시간이 필요합니다. [프로젝트 실행·잠금 파일 안내](https://docs.astral.sh/uv/guides/projects/)를 참고합니다.

### 빈 폴더에서 만들기 — 강사 시연

환경이 어떻게 만들어지는지 보여줄 때만 다음 흐름을 시연합니다. 수강생은 위의 배포 폴더 경로를 사용합니다.

```powershell
mkdir kpc-finance
cd kpc-finance
uv init --bare --python 3.12
uv python pin 3.12
uv add pandas notebook ipykernel openpyxl numpy matplotlib seaborn scikit-learn beautifulsoup4 selenium yfinance
uv run jupyter notebook
```

`uv python pin 3.12`는 실행할 Python 계열을 지정합니다. `uv add`는 필요한 패키지와 잠금 파일을 함께 관리합니다. 강사는 만든 `pyproject.toml`·`uv.lock`·`.python-version`을 검증한 뒤 배포하고 수강생은 같은 파일을 사용합니다. 매번 최신 패키지로 다시 `uv add`하지 않습니다. `.venv` 폴더를 다른 PC로 복사하는 대신 설정과 잠금 파일로 각 PC에서 환경을 만듭니다.

### Notebook 생성과 Kernel 확인

브라우저의 Notebook 화면에서 새 Notebook을 만들고 Python Kernel을 선택합니다. 이름을 `01-python-basics.ipynb`로 저장합니다. 첫 코드 셀을 **Shift+Enter**로 실행합니다. [uv와 Jupyter 공식 안내](https://docs.astral.sh/uv/guides/integration/jupyter/)처럼 프로젝트 환경과 Kernel의 연결을 확인합니다.

````python
print("수업 준비 완료")
print(1 + 2)
````

출력은 `수업 준비 완료`와 `3`입니다. 다음 셀에서 Python의 위치와 패키지 버전을 확인합니다.

````python
import sys
from pathlib import Path
import pandas as pd

print("Python:", sys.version.split()[0])
print("실행 파일:", sys.executable)
print("pandas:", pd.__version__)
print("자료 폴더:", Path.cwd())
````

Windows의 실행 파일은 수업 폴더 아래 `.venv\Scripts\python.exe`인지 확인합니다. 자료를 읽는 기준 위치는 `raw` 폴더가 있는 최상위 수업 폴더입니다. 다른 Kernel이면 서버를 종료하고 수업 폴더에서 `uv run --locked jupyter notebook`으로 다시 엽니다. 패키지를 추가해야 할 때도 Notebook의 `!pip install` 대신 서버를 멈추고 PowerShell에서 강사가 안내한 `uv add 패키지이름`을 사용한 뒤 새 환경을 검증·배포합니다.

Notebook의 변수는 **실행한 셀의 순서**에 따라 저장됩니다. 아래에서는 셀 A를 먼저 실행하고 셀 B를 실행합니다.

셀 A:

````python
base_price = 10000
````

셀 B:

````python
print(base_price + 500)
````

예상 출력은 `10500`입니다. Kernel을 재시작하면 변수 값이 지워지므로 A부터 다시 실행해야 합니다. 이 동작은 [Jupyter 코드 실행 설명](https://jupyter-notebook.readthedocs.io/en/stable/examples/Notebook/Running%20Code.html)에서 확인할 수 있습니다.

**직접 실습:** 출력 문구를 자신의 문구로 바꾸고 실행하세요. 파일을 저장한 뒤 Kernel을 재시작하고 셀을 위에서부터 실행하여 같은 결과가 나오는지 확인하세요.

## 화면에서 구분할 세 가지

**PowerShell**에는 `uv sync --locked`와 `uv run --locked jupyter notebook` 같은 실행 명령을 넣습니다. **코드 셀**에는 `print(...)`처럼 Python 코드를 넣습니다. **Markdown 셀**에는 제목과 수업 메모를 씁니다. 같은 글자를 넣어도 어디에 넣었는지에 따라 처리 방식이 달라집니다.

Notebook의 셀을 클릭하면 입력 내용을 수정할 수 있습니다. Shift+Enter는 현재 셀을 실행하고 다음 셀로 이동합니다. 코드 셀 마지막 줄에 값이나 변수 이름만 적으면 Notebook이 그 값을 보여줍니다. `print()`는 여러 줄을 명시적으로 출력할 때 사용합니다.

Markdown 셀로 바꾼 뒤 아래 문장을 입력하고 실행합니다. 메뉴 이름은 Notebook 버전에 따라 조금 다를 수 있으므로 셀 유형 선택 메뉴에서 Markdown을 고릅니다.

```markdown
# 나의 첫 분석 노트

오늘은 Python 코드를 실행하고 결과를 저장합니다.
```

제목은 큰 글씨로, 설명은 일반 문장으로 표시됩니다. 셀 유형을 다시 Code로 바꿀 때에는 기존 설명이 Python 코드로 실행되지 않도록 새 셀을 사용합니다.

## 저장된 파일과 Kernel의 기억

저장하는 `.ipynb` 파일에는 코드·설명·출력 등이 들어갑니다. 현재 실행 중인 변수는 Kernel의 메모리에 있습니다. 파일을 저장했다고 변수를 만드는 셀을 실행할 필요가 없어지는 것은 아닙니다.

1. `base_price = 10000` 셀을 실행합니다.
2. `print(base_price + 500)` 셀을 실행하여 `10500`을 확인합니다.
3. Kernel 재시작을 선택합니다.
4. 두 번째 셀만 실행하여 `NameError`를 관찰합니다.
5. 첫 셀부터 다시 실행하여 결과가 복구되는지 확인합니다.

이 실험은 오류를 일부러 경험하는 활동입니다. 오류를 확인한 뒤 마지막에는 위에서부터 전체 실행하여 오류 출력이 남지 않게 저장합니다.

## 흔한 오류와 복구

| 현상 | 원인 확인과 복구 |
| --- | --- |
| `uv`가 인식되지 않음 | uv 설치 후 PowerShell을 새로 열고 `uv --version`을 확인합니다. |
| Notebook 셀에서 `uv sync`가 오류 | PowerShell 명령을 코드 셀에 입력했습니다. 수업 폴더의 PowerShell에서 실행합니다. |
| `ModuleNotFoundError: pandas` | `uv sync --locked` 후 `uv run --locked jupyter notebook`으로 실행했고 `sys.executable`이 수업 `.venv`인지 확인합니다. |
| 셀에 문장만 보이고 결과가 없음 | Code/Markdown 유형과 Shift+Enter 실행 여부를 확인합니다. |
| `NameError` | 변수 정의 셀부터 다시 실행합니다. 재시작하면 이전 변수는 사라집니다. |
| 실행 중 표시가 계속됨 | 실행 중인 작업을 확인하고 Kernel의 실행 중단 메뉴를 사용합니다. 필요한 경우 재시작 후 위에서부터 실행합니다. |
| 파일을 어디에 저장했는지 모름 | `Path.cwd()`와 Notebook 파일 목록에서 파일 이름을 확인합니다. |
